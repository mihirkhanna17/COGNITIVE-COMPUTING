```python
# 6.1 Function for the sum of odd values below n
def sum_odd(limit):
    return sum(range(1, limit, 2))

n = 50
print("enter number:", n)
print(sum_odd(n))
```

```python
# 6.2 Function for the sum of prime values up to n
def check_prime(value):
    if value < 2:
        return False
    return all(value % divisor for divisor in range(2, int(value ** 0.5) + 1))

n = 50
prime_total = sum(value for value in range(2, n + 1) if check_prime(value))
print("enter number :", n)
print(prime_total)
```
