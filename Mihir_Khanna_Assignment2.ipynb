```python
# Q1. List operations using the roll number
roll_no = "1024170228"
marks = [int(digit) * 10 for digit in roll_no]

print("Original List:", marks)
marks += [55]
print("After append:", marks)
marks = marks[:3] + [99] + marks[3:]
print("After insert:", marks)
marks.remove(99)
print("After remove:", marks)
marks.pop()
print("After pop:", marks)

ascending = sorted(marks)
descending = sorted(marks, reverse=True)
print("Ascending:", ascending)
print("Descending:", descending)
print("First 3:", descending[:3])
print("Last 3:", descending[-3:])

average = sum(descending) / len(descending)
print("Average =", average)
print("Elements greater than average:", [x for x in descending if x > average])
```

```python
# Q2. Tuple operations
scores = tuple(descending[:8])
print("Tuple:", scores)
highest = max(scores)
lowest = min(scores)
print("Highest Score:", highest)
print("Highest Score Index:", scores.index(highest))
print("Lowest Score:", lowest)
print("Lowest Score Count:", scores.count(lowest))
print(list(scores[::-1]))

search_value = 50
print("Enter score:", search_value)
if search_value in scores:
    print("Index =", scores.index(search_value))
else:
    print("Score not present")

try:
    scores[0] = 100
except TypeError as err:
    print(err)

first, second, *remaining = scores
print(first)
print(second)
print(remaining)
```

```python
# Q3. Generate 100 random integers and analyse them
import random
from collections import Counter

random.seed(1024170228)
values = [random.randint(100, 900) for _ in range(100)]
print(values)

odd_count = sum(v % 2 for v in values)
print("Odd Count:", odd_count)
print("Even Count:", len(values) - odd_count)

def prime(value):
    if value < 2:
        return False
    return all(value % d for d in range(2, int(value ** 0.5) + 1))

primes = [v for v in values if prime(v)]
print("Prime Count:", len(primes))
print("Prime Numbers:", primes)

frequency = Counter(values)
most_common, count = frequency.most_common(1)[0]
print("Most Frequent:", most_common)
print("Frequency:", count)
```

```python
# Q4. Set operations based on the roll number digits
digits = [int(d) for d in roll_no]
A = {d * 7 for d in digits}
B = {d * 9 for d in digits}

print("A =", A)
print("B =", B)
print("Union:", A | B)
print("Intersection:", A & B)
print("A-B:", A - B)
print("B-A:", B - A)
print("Symmetric Difference:", A ^ B)
print("A subset of B:", A <= B)
print("B superset of A:", B >= A)

remove_value = 42
A.discard(remove_value)
print("After discard:", A)
```
