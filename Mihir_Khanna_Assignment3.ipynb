```python
# Q6. Employee dataset
employee_data = {
    "Employee_ID": [101,102,103,104,105],
    "Name": ["Alice","Bob","Charlie","Diana","Edward"],
    "Department": ["HR","IT","IT","Marketing","Sales"],
    "Age": [29,34,41,28,38],
    "Salary": [50000,70000,65000,55000,60000],
    "Years_of_Experience": [4,8,10,3,12],
    "Joining_Date": ["2020-03-15","2017-07-19","2013-06-01","2021-02-10","2010-11-25"],
    "Gender": ["Female","Male","Male","Female","Male"],
    "Bonus": [5000,7000,6000,4500,5000],
    "Rating": [4.5,4.0,3.8,4.7,3.5]
}

employees = pd.DataFrame(employee_data)
employees.to_csv("employees.csv", index=False)
print(employees)
```

```python
# 6(a-l). Analyse and modify the employee data
emp = pd.read_csv("employees.csv")

print(emp.shape)
emp.info()
print(emp.describe())

print("First 5 rows")
print(emp.head())
print("\nLast 3 rows")
print(emp.tail(3))

print("Average Salary =", emp["Salary"].mean())
print("Total Bonus =", emp["Bonus"].sum())
print("Youngest Age =", emp["Age"].min())
print("Highest Rating =", emp["Rating"].max())

ordered = emp.sort_values("Salary", ascending=False)
print(ordered)

def rating_group(value):
    return "Excellent" if value >= 4.5 else ("Good" if value >= 4.0 else "Average")

emp["Performance"] = emp["Rating"].map(rating_group)
print(emp)

print(emp.isna())
print("\nMissing values in each column")
print(emp.isna().sum())

emp = emp.rename(columns={"Employee_ID": "ID"})
print(emp.head())

print(emp.loc[emp["Years_of_Experience"] > 5])
print(emp.loc[emp["Department"].eq("IT")])

emp["Tax"] = emp["Salary"].mul(0.10)
print(emp)

emp.to_csv("employees_modified.csv", index=False)
print("Modified dataset saved successfully.")
```
